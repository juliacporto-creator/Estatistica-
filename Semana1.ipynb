```python
import pandas as pd

# Load the data from the xlsx file into a pandas DataFrame
df = pd.read_excel('/content/dados.xlsx')

# Display the entire DataFrame
display(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get the value counts for the 'Qual dessas frutas você prefere para suco natural?' column
fruit_preferences = df['Qual dessas frutas você prefere para suco natural?'].value_counts().reset_index()
fruit_preferences.columns = ['Fruit', 'Count']

# Create the bar chart with a different palette (e.g., 'plasma')
plt.figure(figsize=(10, 6))
sns.barplot(x='Fruit', y='Count', data=fruit_preferences, palette='plasma', hue='Fruit', legend=False)
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Preferências')
plt.xticks(rotation=72, ha='right')
plt.tight_layout()
plt.show()
```
