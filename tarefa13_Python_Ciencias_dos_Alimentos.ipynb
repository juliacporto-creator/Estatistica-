#Distribuição Normal (Gaussiana)
A distribuição Normal é um modelo de probabilidade para uma variável aleatória contínua, caracterizado por dois parâmetros: a média (μ), que determina a localização central, e a variância (σ²), que representa a dispersão.

O modelo da função densidade pode ser expresso matemáticamente da seguinte forma e permite representar a distribuição dos valores de uma variável contínua em torno de sua média:

Seja X  uma variável aleatória contínua com média μ em que −∞<x<∞
, e σ>0

E:

x = valor da variável aleatória;

μ = média da distribuição;

σ = desvio-padrão da distribuição;

σ²= variância;

e = número de Euler;

π = constante matemática aproximadamente igual a 3,1416.

#Aplicação da distribuição normal na base de dados:

Na base de dados a distribuição Normal pode ser utilizada para investigar uma variável quantitativa contínua, como o pH dos vinhos brancos.

Nesse contexto, uma medição individual corresponde ao valor de pH observado em uma determinada amostra de vinho. Já a média amostral e o desvio-padrão amostral são estatísticas calculadas a partir das amostras de vinhos analisados. Assim, a distribuição Normal não representa uma medição individual, mas sim o  comportamento dos valores de pH da amostra.

A média μ pode assumir qualquer valor real e determina a posição central da curva. Quando a média aumenta, a curva se desloca para a direita; quando diminui, desloca-se para a esquerda. O desvio-padrão σ deve ser positivo e determina a dispersão dos valores em torno da média. Valores menores de σ produzem uma curva mais concentrada e alta, enquanto valores maiores produzem uma curva mais espalhada e baixa.

```python
ph_mean = df['pH'].mean()
ph_std = df['pH'].std()

print(f"Média do pH: {ph_mean:.2f}")
print(f"Desvio Padrão do pH: {ph_std:.2f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Set the range for pH values for plotting
# We'll go from min to max pH in the dataset, or a bit wider around the mean
# for a smoother curve, using 4 standard deviations on each side for a good range.
min_ph = df['pH'].min()
max_ph = df['pH'].max()

# Create a range of values for the x-axis (pH values)
x = np.linspace(min_ph - 0.5, max_ph + 0.5, 1000)

# Calculate the probability density function (PDF) for each x value
pdf = norm.pdf(x, ph_mean, ph_std)

# Plot the normal distribution curve
plt.figure(figsize=(12, 7))

# Plot the histogram of pH values
plt.hist(df['pH'], bins=30, density=True, alpha=0.6, color='green', label='Histograma de pH')

# Plot the normal distribution curve
plt.plot(x, pdf, label=f'Distribuição Normal (μ={ph_mean:.2f}, σ={ph_std:.2f})', color='blue', linewidth=2)

plt.title('Distribuição Normal e Histograma do pH do Vinho Branco')
plt.xlabel('pH')
plt.ylabel('Densidade de Probabilidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

Para analisar o efeito do parâmetro μ sobre a distribuição Normal, agora consideramos o valor hipotético de 3,4 média, mantendo o desvio-padrão constante

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Define a nova média hipotética
ph_mean_hypothetical = 3.4

# Use o desvio padrão original
# ph_std já foi calculado anteriormente e está disponível no ambiente

# Create a range of values for the x-axis (pH values)
# Reutilizando min_ph e max_ph para manter a consistência do range do gráfico
x_hypothetical = np.linspace(min_ph - 0.5, max_ph + 0.5, 1000)

# Calculate the probability density function (PDF) for the hypothetical mean
pdf_hypothetical = norm.pdf(x_hypothetical, ph_mean_hypothetical, ph_std)

# Plot the normal distribution curve with the hypothetical mean and the original histogram
plt.figure(figsize=(12, 7))

# Plot the histogram of pH values (from the original data)
plt.hist(df['pH'], bins=30, density=True, alpha=0.6, color='green', label='Histograma de pH Original')

# Plot the normal distribution curve with the hypothetical mean
plt.plot(x_hypothetical, pdf_hypothetical, label=f'Distribuição Normal (μ={ph_mean_hypothetical:.2f}, σ={ph_std:.2f})', color='red', linewidth=2)

# Optionally, plot the original normal distribution for comparison
# plt.plot(x, pdf, label=f'Distribuição Normal Original (μ={ph_mean:.2f}, σ={ph_std:.2f})', color='blue', linestyle='--', linewidth=1)

plt.title(f'Distribuição Normal do pH com Média Hipotética ({ph_mean_hypothetical:.2f}) e Histograma Original')
plt.xlabel('pH')
plt.ylabel('Densidade de Probabilidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

Para analisar o efeito do parâmetro σ sobre a distribuição Normal, agora consideramos o valor hipotético de 0,3 de desvio padrão, mantendo a média constante.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Use a média original
# ph_mean já foi calculado anteriormente e está disponível no ambiente

# Define o novo desvio padrão hipotético
ph_std_hypothetical = 0.3

# Create a range of values for the x-axis (pH values)
x_hypothetical = np.linspace(min_ph - 0.5, max_ph + 0.5, 1000)

# Calculate the probability density function (PDF) for the hypothetical standard deviation
pdf_hypothetical = norm.pdf(x_hypothetical, ph_mean, ph_std_hypothetical)

# Plot the normal distribution curve with the hypothetical standard deviation and the original histogram
plt.figure(figsize=(12, 7))

# Plot the histogram of pH values (from the original data)
plt.hist(df['pH'], bins=30, density=True, alpha=0.6, color='green', label='Histograma de pH Original')

# Plot the normal distribution curve with the hypothetical standard deviation
plt.plot(x_hypothetical, pdf_hypothetical, label=f'Distribuição Normal (μ={ph_mean:.2f}, σ={ph_std_hypothetical:.2f})', color='purple', linewidth=2)

# Optionally, plot the original normal distribution for comparison
# plt.plot(x, pdf, label=f'Distribuição Normal Original (μ={ph_mean:.2f}, σ={ph_std:.2f})', color='blue', linestyle='--', linewidth=1)

plt.title(f'Distribuição Normal do pH com Desvio Padrão Hipotético ({ph_std_hypothetical:.2f}) e Histograma Original')
plt.xlabel('pH')
plt.ylabel('Densidade de Probabilidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

Com a criação dos três gráficos acima é possível exemplificar as mudanças já citadas.

Ao variar a média μ e manter o desvio-padrão σ constante, a distribuição Normal sofre um deslocamento horizontal. O aumento de μ desloca a curva para a direita, enquanto a redução de μ desloca a curva para a esquerda. A dispersão e a simetria da distribuição permanecem inalteradas. Portanto, a média determina principalmente a posição da curva Normal.

Ao variar o desvio-padrão σ e manter a média μ constante, há alteração na dispersão da distribuição Normal. Quando σ aumenta, os valores ficam mais espalhados em torno da média, produzindo uma curva mais larga e com menor altura. Quando σ diminui, os valores ficam mais concentrados, resultando em uma curva mais estreita e alta. A posição da curva permanece constante e a distribuição continua simétrica. Portanto, o desvio-padrão determina principalmente a dispersão dos valores em torno da média.

Nenhuma das duas alterações produz assimetria. A distribuição Normal permanece simétrica independentemente dos valores de μ e σ, desde que σ > 0.

###Condições e limitações:

A aplicação da distribuição normal aos valores de pH pressupõe que os dados apresentem comportamento aproximadamente simétrico e possam ser adequadamente representados por uma distribuição em forma de sino.

Neste estudo, foram utilizados a média de 3,19 e o desvio-padrão de 0,15 para caracterizar a distribuição dos valores de pH. A variação da média desloca a distribuição para a direita ou para a esquerda, enquanto a variação do desvio-padrão altera sua dispersão, tornando a curva mais estreita ou mais ampla. Dessa forma, as probabilidades calculadas dependem diretamente desses dois parâmetros.

Como limitação, destaca-se que a distribuição normal é uma aproximação dos dados observados e que a média e o desvio-padrão foram obtidos a partir da amostra, podendo não representar perfeitamente uma população mais ampla de vinhos. Além disso, eventuais assimetrias ou valores extremos presentes na base podem reduzir a adequação do modelo normal.

#Probabilidades com Phyton - Distribuição Normal

##Probabilidade acumulada

Qual é a probabilidade de um vinho selecionado aleatoriamente apresentar pH menor ou igual a 3,10?

Aqui estamos calculando:

$$ P(X \leq 3,10) $$

É uma probabilidade acumulada, porque considera todos os valores de pH desde o menor valor observado até 3,10.

##Probabilidade entre dois valores

Qual é a probabilidade de um vinho selecionado aleatoriamente apresentar pH entre 3,10 e 3,30?

Aqui:

$$ P(3,10 \leq X \leq 3,30) $$

É uma probabilidade entre dois valores, pois queremos a área da curva compreendida entre 3,10 e 3,30.

##Probabilidade acima de um valor

Qual é a probabilidade de um vinho selecionado aleatoriamente apresentar pH superior a 3,40?

Aqui:

$$ P(X > 3,40) $$

É uma probabilidade acima de um valor, correspondendo à área da curva à direita de 3,40.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Recalculate probabilities (or use the already calculated variables)
prob_le_310 = norm.cdf(3.10, loc=ph_mean, scale=ph_std)
prob_le_330 = norm.cdf(3.30, loc=ph_mean, scale=ph_std)
prob_between_310_330 = prob_le_330 - prob_le_310
prob_gt_340 = 1 - norm.cdf(3.40, loc=ph_mean, scale=ph_std)

# Create a range of x values for the plot
x_plot = np.linspace(ph_mean - 4 * ph_std, ph_mean + 4 * ph_std, 1000)
y_plot = norm.pdf(x_plot, loc=ph_mean, scale=ph_std)

plt.figure(figsize=(12, 7))
plt.plot(x_plot, y_plot, color='blue', label=f'Distribuição Normal (μ={ph_mean:.2f}, σ={ph_std:.2f})')

# Shade the area P(X <= 3.10)
x_fill_le = x_plot[x_plot <= 3.10]
y_fill_le = y_plot[x_plot <= 3.10]
plt.fill_between(x_fill_le, y_fill_le, color='skyblue', alpha=0.5, label=f'P(pH ≤ 3.10) = {prob_le_310:.4f}')

# Shade the area P(3.10 <= X <= 3.30)
x_fill_between = x_plot[(x_plot >= 3.10) & (x_plot <= 3.30)]
y_fill_between = y_plot[(x_plot >= 3.10) & (x_plot <= 3.30)]
plt.fill_between(x_fill_between, y_fill_between, color='lightcoral', alpha=0.5, label=f'P(3.10 ≤ pH ≤ 3.30) = {prob_between_310_330:.4f}')

# Shade the area P(X > 3.40)
x_fill_gt = x_plot[x_plot > 3.40]
y_fill_gt = y_plot[x_plot > 3.40]
plt.fill_between(x_fill_gt, y_fill_gt, color='lightgreen', alpha=0.5, label=f'P(pH > 3.40) = {prob_gt_340:.4f}')

plt.title('Probabilidades de pH do Vinho Branco na Distribuição Normal')
plt.xlabel('pH')
plt.ylabel('Densidade de Probabilidade')

# Add vertical lines for better visualization
plt.axvline(3.10, color='gray', linestyle=':', linewidth=0.8)
plt.axvline(3.30, color='gray', linestyle=':', linewidth=0.8)
plt.axvline(3.40, color='gray', linestyle=':', linewidth=0.8)

plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

##Probabilidade de pH ≤ 3.10

com a área sombreada em azul claro representando essa probabilidade. O valor calculado é de P(pH ≤ 3.10) = 0.2794.

A área sombreada em azul claro corresponde aproximadamente a 28% da área total sob a curva normal, que se estende de um pH muito baixo até 3.10. Isso representa a probabilidade de um vinho ter um pH igual ou inferior a 3.10.

Esse limite foi escolhido pois 3,10 está cerca de 0,60 desvio-padrão abaixo da média, então não se espera uma probabilidade muito pequena.

##Probabilidade de pH entre 3.10 e 3.30

com a área sombreada em coral claro. O valor calculado é de P(3.10 ≤ pH ≤ 3.30) = 0.4909.

A área em coral claro, localizada entre os valores de pH de 3.10 e 3.30, representa quase 50% da área total sob a curva. É a maior das três áreas sombreadas, indicando a maior probabilidade.

Esses limites foram escolhidos pois esse intervalo contém a média (3,19), representando uma região central da distribuição.


##Probabilidade de pH > 3.40

com a área sombreada em verde claro. O valor calculado é de P(pH > 3.40) = 0.0804.

A área em verde claro, que se estende de 3.40 para a direita, é a menor das três, representando cerca de 8% da área total. Esta porção da curva mostra a probabilidade de um vinho ter um pH superior a 3.40.

Esse limite foi escolhido pois 3,40 está aproximadamente 1,4 desvios-padrão acima da média, então teremos uma área menor na cauda direita.

## Aplicação da distribuição de t de Student na base de dados

A distribuição t de Student pode representar a variabilidade de uma estatística amostral, especialmente a média de uma variável, como o teor de álcool, entre diferentes amostras de vinhos, a distribuição da estatística de teste associada à média amostral, principalmente quando o desvio-padrão populacional é desconhecido

A medição individual, nesse contexto, corresponde ao valor de teor de álcool de cada vinho. Enquanto a estatística calculada é o valor obtido de várias observações

O parâmetro de grau de liberdade representados pela letra grega ν (ni) assume valores reais positivos na definição matemática da distribuição que controla o formato da curva. Valores menores produzem caudas mais pesadas e menor concentração central, valores maiores aproximam a curva da distribuição normal padrão.

A variável aleatória t pode assumir qualquer valor real, entre menos e mais infinito, e representa os valores que a função densidade é avaliada. Ela não é um parâmetro que controla o formato da curva, mas corresponde ao eixo horizontal do gráfico.

A função gama, representada por Γ, também aparece na fórmula da distribuição. Ela é uma função matemática utilizada no cálculo da constante de normalização, garantindo que a área total sob a curva seja igual a 1.

A constante π, aproximadamente igual a 3,14159, também participa desse cálculo. Tanto a função gama quanto a constante π fazem parte da expressão matemática e não são parâmetros livres da distribuição.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# Gerar pontos para o eixo x
x_range = np.linspace(-4, 4, 1000)

# Configurar o tamanho da figura
plt.figure(figsize=(10, 6))

# Plotar a distribuição t de Student para diferentes graus de liberdade (gl)
for gl in [1, 2, 5, 30]:
    plt.plot(x_range, t.pdf(x_range, df=gl), label=f't de Student (gl = {gl})', linewidth=2)

# Plotar a Distribuição Normal Padrão para fins de comparação
from scipy.stats import norm
plt.plot(x_range, norm.pdf(x_range), 'r--', label='Normal Padrão (N(0,1))', linewidth=2)

# Customizações do gráfico
plt.title('Curvas da Distribuição t de Student vs. Normal Padrão')
plt.xlabel('Desvios Padrão (t)')
plt.ylabel('Densidade de Probabilidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

Representa a densidade de probabilidade da distribuição t para diferentes graus de liberdade.

O gráfico como objetivo mostrar como a distribuição muda conforme os graus de liberdade aumentam. Com muitos graus de liberdade, ela se aproxima da distribuição normal padrão.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# Vamos definir os graus de liberdade baseados no Welch do teste anterior
# Ou um valor padrão representativo, por exemplo, gl = 30
gl = 30
confianca = 0.95
alpha = 1 - confianca

# Encontrar o t crítico para teste bicaudal
t_critico_esq = t.ppf(alpha / 2, df=gl)
t_critico_dir = t.ppf(1 - alpha / 2, df=gl)

# Gerar valores de t para o eixo x
x_vals = np.linspace(-4, 4, 1000)
y_vals = t.pdf(x_vals, df=gl)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, color='darkblue', linewidth=2, label=f'Distribuição t de Student (gl = {gl})')

# Sombrear a área de rejeição (região crítica)
x_dir = x_vals[x_vals >= t_critico_dir]
y_dir = y_vals[x_vals >= t_critico_dir]
plt.fill_between(x_dir, y_dir, color='tomato', alpha=0.6, label=f'Região Crítica Superior (t ≥ {t_critico_dir:.3f})')

x_esq = x_vals[x_vals <= t_critico_esq]
y_esq = y_vals[x_vals <= t_critico_esq]
plt.fill_between(x_esq, y_esq, color='tomato', alpha=0.6, label=f'Região Crítica Inferior (t ≤ {t_critico_esq:.3f})')

# Linhas verticais do t crítico
plt.axvline(t_critico_dir, color='red', linestyle='--', linewidth=1.5, label=f't Crítico = ±{t_critico_dir:.3f}')
plt.axvline(t_critico_esq, color='red', linestyle='--', linewidth=1.5)

# Customização do layout
plt.title(f'Distribuição t de Student com Limites de t Crítico (Confiança: {confianca*100:.0f}%, gl = {gl})', fontsize=12, fontweight='bold')
plt.xlabel('Valor t', fontsize=11)
plt.ylabel('Densidade de Probabilidade', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.show()
```

##Interpretação do gráfico

1. O Formato da Curva (Linha Azul Escuro): A curva representa a densidade de probabilidade da estatística t. Ela é muito parecida com a curva Normal Padrão, mas possui caudas ligeiramente mais pesadas  devido ao menor número de graus de liberdade (gl = 30).

2. O t Crítico (Linhas Vermelhas Tracejadas em $\pm 2,042$): Os valores de t crítico funcionam como as barreiras ou fronteiras de decisão.
Esses valores são definidos a partir do nível de significância escolhido ($\alpha = 0,05$ ou 5% no teste bicaudal, o que deixa 2,5% de probabilidade em cada cauda).

3. As Áreas Sombreadas em Vermelho/Laranja (Região Crítica ou de Rejeição): As duas áreas nas extremidades (cauda esquerda e cauda direita) representam a Região Crítica (Região de Rejeição) da hipótese nula ($H_0$).
Se você realizar um teste estatístico (como um teste t de comparação de médias) e o valor de $t$ calculado a partir dos seus dados for maior que $2,042$ ou menor que $-2,042$, significa que a sua estatística caiu na região vermelha.
Conclusão estatística nessa situação: Rejeita-se a hipótese nula ($H_0$), indicando que há uma diferença estatisticamente significativa entre os grupos analisados.

4. A Área Branca Central (Região de Não Rejeição)
Representa a maior parte da curva (95% de probabilidade): Se o valor de $t$ calculado a partir dos dados do seu teste de hipótese cair dentro deste intervalo (entre $-2,042$ e $+2,042$), você não rejeita a hipótese nula ($H_0$), concluindo que as diferenças observadas podem ser atribuídas apenas ao acaso.

## Condições e Limitações

Para que as inferências realizadas por meio da distribuição t de Student sejam matematicamente válidas, algumas condições devem ser satisfeitas:

- Independência das Observações: Cada medição ou amostra de vinho na base de dados deve ser independente das demais. No contexto deste conjunto de dados, pressupõe-se que a seleção de uma garrafa de vinho não influencia as características físico-químicas de outra.
- Normalidade da População: Teoricamente, a distribuição t assume que a variável de interesse (como o teor alcoólico ou pH) segue uma distribuição normal na população de origem.
- Homocedasticidade (Variâncias Iguais): Para o teste t padrão de duas amostras, assume-se que os dois grupos comparados possuem variâncias semelhantes. Caso essa premissa seja violada, utiliza-se a correção de Welch (como exemplificado nos graus de liberdade fracionados), que ajusta os graus de liberdade para corrigir a heterocedasticidade.

Eventuais limitações:
- Presença de Outliers e Erros de Escala: Conforme observado nas análises prévias do conjunto de dados, a presença de ruídos ou valores extremos (outliers) pode distorcer severamente o cálculo da média amostral e aumentar artificialmente o desvio-padrão amostral, afetando o erro padrão e afastando as estimativas da realidade populacional.
- Viés de Amostragem: Os dados provêm especificamente de amostras de Vinho Verde Português branco. Portanto, qualquer generalização ou inferência estatística usando a distribuição t fica limitada a este nicho de mercado e região produtora, não sendo estatisticamente correto extrapolar as conclusões para vinhos tintos ou vinhos produzidos em outras regiões geográficas globais.
- Amostras com Tamanhos Muito Desequilibrados: Se dividirmos as análises em grupos baseados em notas de qualidade muito raras (como notas extremamente baixas ou extremamente altas), teremos subgrupos com pouquíssimas amostras ($n \le 5$), onde a sensibilidade a desvios de normalidade aumenta significativamente e a margem de erro do $t$ crítico se torna muito ampla.



Suponha que estamos analisando o teor alcoólico médio de um novo lote de vinhos verdes brancos. Coletamos uma amostra de **16 vinhos** (n = 16), obtendo uma média amostral de 10,4% e um desvio-padrão amostral de 1,2%.

Queremos testar a hipótese de que o teor alcoólico médio populacional seja realmente $\mu_0 = 10,0\%$. Para isso, a nossa estatística de teste padronizada segue a distribuição t de Student com $v = n - 1 = 15$ graus de liberdade:

$$ t = \frac{\bar{X} - \mu_0}{S / \sqrt{n}} $$


1. **Probabilidade Acumulada (Inferior):** Qual é a probabilidade de obtermos um valor de $t$ menor ou igual a $-1,5$?
   $$ P(t \leq -1,5) $$
   *Significado do limite:* Representa o cenário onde a média do lote sob teste está substancialmente abaixo da hipótese de 10,0%.

2. **Probabilidade Entre Dois Valores:** Qual é a probabilidade de o valor de $t$ estar entre $-1,5$ e $1,5$?
   $$ P(-1,5 \leq t \leq 1,5) $$
   *Significado dos limites:* Representa a faixa de valores centrais onde a média amostral é considerada estatisticamente próxima de 10,0%.

3. **Probabilidade Acima de um Valor:** Qual é a probabilidade de obtermos um valor de $t$ superior a $2,0$?
   $$ P(t > 2,0) $$
   *Significado do limite:* Representa um cenário atípico de teor alcoólico médio significativamente superior ao esperado de 10,0%.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# Definição dos graus de liberdade
gl_amostra = 15

# 1. Probabilidade acumulada P(t <= -1.5)
prob_acumulada = t.cdf(-1.5, df=gl_amostra)

# 2. Probabilidade entre -1.5 e 1.5
prob_entre = t.cdf(1.5, df=gl_amostra) - t.cdf(-1.5, df=gl_amostra)

# 3. Probabilidade acima de 2.0
prob_acima = 1 - t.cdf(2.0, df=gl_amostra)

print(f"1. P(t <= -1.5) = {prob_acumulada:.4f} (ou {prob_acumulada*100:.2f}%)")
print(f"2. P(-1.5 <= t <= 1.5) = {prob_entre:.4f} (ou {prob_entre*100:.2f}%)")
print(f"3. P(t > 2.0) = {prob_acima:.4f} (ou {prob_acima*100:.2f}%)")
```

```python
# Gerar valores para o eixo x
x_t = np.linspace(-4, 4, 1000)
y_t = t.pdf(x_t, df=gl_amostra)

plt.figure(figsize=(12, 7))
plt.plot(x_t, y_t, color='blue', linewidth=2, label=f't de Student (gl = {gl_amostra})')

# Sombrear P(t <= -1.5)
x_fill_inf = x_t[x_t <= -1.5]
y_fill_inf = y_t[x_t <= -1.5]
plt.fill_between(x_fill_inf, y_fill_inf, color='skyblue', alpha=0.5, label=f'P(t ≤ -1.5) = {prob_acumulada:.4f}')

# Sombrear P(-1.5 <= t <= 1.5)
x_fill_mid = x_t[(x_t >= -1.5) & (x_t <= 1.5)]
y_fill_mid = y_t[(x_t >= -1.5) & (x_t <= 1.5)]
plt.fill_between(x_fill_mid, y_fill_mid, color='lightcoral', alpha=0.5, label=f'P(-1.5 ≤ t ≤ 1.5) = {prob_entre:.4f}')

# Sombrear P(t > 2.0)
x_fill_sup = x_t[x_t > 2.0]
y_fill_sup = y_t[x_t > 2.0]
plt.fill_between(x_fill_sup, y_fill_sup, color='lightgreen', alpha=0.5, label=f'P(t > 2.0) = {prob_acima:.4f}')

# Adicionando detalhes ao gráfico
plt.title('Áreas de Probabilidade na Distribuição t de Student (gl = 15)', fontsize=14, fontweight='bold')
plt.xlabel('Valor t')
plt.ylabel('Densidade de Probabilidade')

plt.axvline(-1.5, color='gray', linestyle=':', alpha=0.7)
plt.axvline(1.5, color='gray', linestyle=':', alpha=0.7)
plt.axvline(2.0, color='gray', linestyle=':', alpha=0.7)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.show()
```

### Comparação do Resultado Numérico com o Gráfico:
- **Área Azul Claro ($P(t \leq -1,5)$):** O cálculo numérico indica **7,71%**. Visualmente, essa cauda esquerda representa uma fração pequena no extremo, condizente com a estimativa matemática.
- **Área Coral ($P(-1,5 \leq t \leq 1,5)$):** Corresponde à maior parte central da distribuição com **84,58%** da área sob a curva. No gráfico, a região coral engloba quase todo o centro e o pico, correspondendo à grande probabilidade calculada.
- **Área Verde Claro ($P(t > 2,0)$):** Reduzida à extremidade direita com valor numérico de **3,20%**. Graficamente, é a menor das três regiões destacadas, mostrando que valores extremos acima de 2 desvios amostrais são improváveis sob a hipótese nula.

##Distribuição de Qui-quadrado
É um teste de hipóteses que se destina a encontrar um valor da dispersão para duas variáveis categóricas nominais e avaliar a associação existente entre variáveis qualitativas. Não depende de parâmetros populacionais (média e variância), o princípio básico deste teste é comparar proporções, ou seja, possíveis divergências entre as frequências observadas e esperadas para um certo evento.

#Aplicação da distribuição normal na base de dados:

No contexto da base de dados de qualidade de vinhos brancos, a distribuição qui-quadrado pode ser utilizada para representar a distribuição de uma estatística relacionada à variância do pH de uma amostra de vinhos. Nesse caso, considera-se uma amostra de tamanho n retirada de uma população de vinhos brancos, cuja variância populacional é σ².

A estatística utilizada é:

χ² = [(n − 1) · S²] / σ²

em que S² representa a variância calculada a partir da amostra.

É importante diferenciar uma medição individual de uma estatística amostral. Por exemplo, o pH de um determinado vinho da base, como 3,20, é uma medição individual. Já a variância S², calculada a partir dos valores de pH de vários vinhos, é uma estatística obtida da amostra. Assim, a distribuição qui-quadrado não descreve diretamente os valores individuais de pH, mas sim o comportamento da estatística relacionada à variância quando diferentes amostras são consideradas.

Dessa forma, a distribuição qui-quadrado permite estudar a variabilidade do pH entre diferentes amostras de vinhos. Para uma amostra de tamanho n, a distribuição possui n − 1 graus de liberdade. Por exemplo, se forem selecionados 20 vinhos, a estatística associada à variância seguirá uma distribuição qui-quadrado com 19 graus de liberdade, desde que sejam atendidas as condições necessárias para essa aplicação.

A distribuição qui-quadrado possui um único parâmetro: os graus de liberdade, representados por k (ou, em algumas fontes, por ν).

Esse parâmetro está diretamente relacionado ao tamanho da amostra. No contexto da variância amostral:

k = n − 1

em que n é o tamanho da amostra.

Por exemplo, se forem analisados 20 vinhos:

k = 20 − 1 = 19

Portanto, a estatística relacionada à variância seguirá uma distribuição qui-quadrado com 19 graus de liberdade.

Valores admissíveis

Para a distribuição qui-quadrado, os graus de liberdade devem ser positivos:

k > 0

Na aplicação estatística associada à variância amostral, como k = n − 1, normalmente k assume valores inteiros positivos.

##Efeito dos graus de liberdade na curva

Os graus de liberdade alteram principalmente a forma, a posição e a assimetria da curva.

Poucos graus de liberdade: a curva apresenta maior assimetria à direita, com uma cauda direita mais alongada.

Aumento dos graus de liberdade: a curva se desloca para a direita, torna-se mais espalhada e sua assimetria diminui.

Muitos graus de liberdade: a distribuição fica progressivamente mais próxima de uma distribuição Normal, embora continue sendo assimétrica e limitada a valores maiores que zero.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Definir o intervalo de valores para o eixo x
x_range = np.linspace(0, 40, 1000)

# Configurar o tamanho do gráfico
plt.figure(figsize=(12, 7))

# Plotar a distribuição Qui-quadrado para diferentes graus de liberdade (k)
graus_liberdade = [2, 5, 10, 20]
cores = ['blue', 'green', 'orange', 'purple']

for k, cor in zip(graus_liberdade, cores):
    plt.plot(x_range, chi2.pdf(x_range, df=k), label=f'Qui-quadrado (k = {k})', color=cor, linewidth=2)

# Customizações do gráfico
plt.title('Distribuição Qui-quadrado para Diferentes Graus de Liberdade (k)', fontsize=14, fontweight='bold')
plt.xlabel('Valor de χ²', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.xlim(0, 40)
plt.ylim(0, 0.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=11)
plt.show()
```

### Análise Estatística do Gráfico da Distribuição Qui-quadrado

A partir do gráfico gerado para os diferentes graus de liberdade ($k = 2, 5, 10$ e $20$), podemos realizar as seguintes observações e análises importantes sobre o comportamento da distribuição Qui-quadrado:

1. **Efeito na Forma e Assimetria:**
   * **Para $k = 2$ (Curva Azul):** A distribuição é extremamente assimétrica à direita, iniciando seu pico logo no início (próximo de zero) e decaindo rapidamente de forma exponencial. Isso mostra que, para poucos graus de liberdade, a maior parte da probabilidade está concentrada em valores muito baixos.
   * **À medida que $k$ aumenta ($5, 10$ e $20$):** A assimetria à direita diminui progressivamente. A curva começa a se achatar e a se alongar, tornando-se mais simétrica em torno do seu pico.

2. **Deslocamento do Pico (Moda e Média):**

3. **Dispersão e Variabilidade:**
   * A variância de uma distribuição Qui-quadrado é dada por $2k$. Portanto, quanto maior o grau de liberdade, maior é a dispersão dos dados.
   * Isso é refletido nas curvas: a curva de $k = 2$ é muito estreita e alta (pouca variabilidade), enquanto a curva de $k = 20$ é muito mais achatada e espalhada (alta variabilidade).

### Condições Necessárias para Aplicação e Limitações do Modelo

Para que a distribuição Qui-quadrado ($\\chi^2$) seja aplicada de forma estatisticamente válida para inferir sobre a variância amostral do pH dos vinhos, certas premissas teóricas devem ser cumpridas, além de considerarmos as limitações práticas dos dados analisados.

#### 1. Condições Necessárias para Aplicação

* **Normalidade da População de Origem (Premissa Crítica):**
  A estatística de teste para a variância, dada por $\\chi^2 = \\frac{(n-1)S^2}{\\sigma^2}$, assume estritamente que a variável sob estudo (o pH) segue uma **Distribuição Normal** na população de origem. Se a distribuição do pH for fortemente assimétrica ou tiver caudas muito pesadas, a distribuição amostral de $S^2$ não seguirá perfeitamente uma distribuição Qui-quadrado, o que pode invalidar testes de hipóteses ou intervalos de confiança construídos sob essa premissa.
  * *No nosso caso:* Como visto na seção da Distribuição Normal, o pH do vinho branco apresenta um comportamento visualmente simétrico e muito próximo a uma curva Gaussiana, o que valida essa condição básica.

* **Independência das Amostras:**
  As observações de pH de cada garrafa de vinho devem ser independentes entre si. A seleção e a medição do pH de uma amostra não podem influenciar ou estar correlacionadas com as de outra amostra (amostragem aleatória simples).

* **Escala de Medida Contínua:**
  A variável analisada (pH) é quantitativa contínua, atendendo ao requisito de que a distribuição Qui-quadrado modela variáveis contínuas derivadas de somas de quadrados de variáveis normais padronizadas.

#### 2. Limitações Práticas e dos Dados

* **Sensibilidade a Outliers:**
  A estatística Qui-quadrado baseia-se na variância amostral ($S^2$), que por sua vez é altamente sensível a valores extremos (*outliers*). Se a base de dados contivesse erros de digitação ou vinhos com pH extremamente discrepantes do padrão (por problemas de medição ou defeitos na fabricação), a variância calculada seria severamente inflacionada, distorcendo a análise do Qui-quadrado.

* **Viés de Amostragem (Generalização Limitada):**
  Os dados são oriundos exclusivamente de amostras de Vinho Verde Português branco. Os resultados e o comportamento da variância obtidos através deste modelo não podem ser extrapolados para outros tipos de vinhos (como vinhos tintos, que possuem faixas de pH e variabilidades tipicamente distintas) ou vinhos produzidos em outras regiões geográficas com diferentes solos e microclimas.

* **Aproximação Amostral de $\\sigma^2$:**
  Em aplicações práticas, a variância populacional real ($\\sigma^2$) é desconhecida e costuma ser estimada ou comparada com valores históricos. A qualidade do teste Qui-quadrado depende diretamente da confiabilidade do parâmetro populacional adotado como referência.

## Exercício Prático e Probabilidades - Distribuição Qui-quadrado

### Cenário Contextualizado

Suponha que um produtor de vinho verde branco queira garantir a consistência do nível de acidez (pH) de seus lotes de engarrafamento. Historicamente, sabe-se que a variância populacional do pH para este tipo de vinho é de $\sigma^2 = 0,0225$ (o que equivale ao desvio-padrão histórico de $\sigma = 0,15$ obtido na nossa análise geral).

Para fins de controle de qualidade, o produtor coleta uma amostra aleatória de **21 vinhos** ($n = 21$), o que nos dá $k = n - 1 = 20$ graus de liberdade para estimar a variância amostral ($S^2$). Sob a premissa de normalidade do pH, a estatística de teste padronizada segue a distribuição Qui-quadrado:

$$ \chi^2 = \frac{(n-1)S^2}{\sigma^2} = \frac{20 \cdot S^2}{0,0225} $$

Formulamos as seguintes perguntas de probabilidade para monitorar essa variabilidade:

1. **Probabilidade Acumulada (Inferior):** Qual é a probabilidade de obtermos um valor de $\chi^2$ menor ou igual a $10,0$?
   $$ P(\chi^2 \le 10,0) $$
   *Significado do limite:* Corresponde a um cenário em que a variância amostral obtida é muito baixa ($S^2 \le 0,01125$), sugerindo um lote excepcionalmente uniforme.

2. **Probabilidade Entre Dois Valores:** Qual é a probabilidade de a estatística $\chi^2$ estar entre $12,0$ e $28,0$?
   $$ P(12,0 \le \chi^2 \le 28,0) $$
   *Significado dos limites:* Representa a faixa de variabilidade considerada ideal e padrão para o controle de qualidade do processo.

3. **Probabilidade Acima de um Valor:** Qual é a probabilidade de obtermos uma estatística $\chi^2$ maior que $32,0$?
   $$ P(\chi^2 > 32,0) $$
   *Significado do limite:* Representa um cenário atípico de alta instabilidade e descontrole de acidez no lote ($S^2 > 0,036$).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Definição dos graus de liberdade
gl_chi = 20

# 1. Probabilidade acumulada P(Chi^2 <= 10.0)
prob_acum_chi = chi2.cdf(10.0, df=gl_chi)

# 2. Probabilidade entre 12.0 e 28.0
prob_entre_chi = chi2.cdf(28.0, df=gl_chi) - chi2.cdf(12.0, df=gl_chi)

# 3. Probabilidade acima de 32.0
prob_acima_chi = 1 - chi2.cdf(32.0, df=gl_chi)

# 4. Cálculo do 90º percentil
percentil_90_chi = chi2.ppf(0.90, df=gl_chi)

print(f"1. P(Chi^2 <= 10.0) = {prob_acum_chi:.4f} (ou {prob_acum_chi*100:.2f}%)")
print(f"2. P(12.0 <= Chi^2 <= 28.0) = {prob_entre_chi:.4f} (ou {prob_entre_chi*100:.2f}%)")
print(f"3. P(Chi^2 > 32.0) = {prob_acima_chi:.4f} (ou {prob_acima_chi*100:.2f}%)")
print(f"O 90º percentil da distribuição Qui-quadrado (gl=20) é: {percentil_90_chi:.3f}")
```

```python
# Gerar valores para o eixo x da distribuição Qui-quadrado
x_c = np.linspace(0, 45, 1000)
y_c = chi2.pdf(x_c, df=gl_chi)

plt.figure(figsize=(12, 7))
plt.plot(x_c, y_c, color='purple', linewidth=2, label=f'Qui-quadrado (gl = {gl_chi})')

# Sombrear P(Chi^2 <= 10.0)
x_fill_inf_chi = x_c[x_c <= 10.0]
y_fill_inf_chi = y_c[x_c <= 10.0]
plt.fill_between(x_fill_inf_chi, y_fill_inf_chi, color='skyblue', alpha=0.5, label=f'P(χ² ≤ 10.0) = {prob_acum_chi:.4f}')

# Sombrear P(12.0 <= Chi^2 <= 28.0)
x_fill_mid_chi = x_c[(x_c >= 12.0) & (x_c <= 28.0)]
y_fill_mid_chi = y_c[(x_c >= 12.0) & (x_c <= 28.0)]
plt.fill_between(x_fill_mid_chi, y_fill_mid_chi, color='lightcoral', alpha=0.5, label=f'P(12.0 ≤ χ² ≤ 28.0) = {prob_entre_chi:.4f}')

# Sombrear P(Chi^2 > 32.0)
x_fill_sup_chi = x_c[x_c > 32.0]
y_fill_sup_chi = y_c[x_c > 32.0]
plt.fill_between(x_fill_sup_chi, y_fill_sup_chi, color='lightgreen', alpha=0.5, label=f'P(χ² > 32.0) = {prob_acima_chi:.4f}')

# Detalhes adicionais do gráfico
plt.title('Áreas de Probabilidade na Distribuição Qui-quadrado (gl = 20)', fontsize=14, fontweight='bold')
plt.xlabel('Valor de χ²')
plt.ylabel('Densidade de Probabilidade')

plt.axvline(10.0, color='gray', linestyle=':', alpha=0.7)
plt.axvline(12.0, color='gray', linestyle=':', alpha=0.7)
plt.axvline(28.0, color='gray', linestyle=':', alpha=0.7)
plt.axvline(32.0, color='gray', linestyle=':', alpha=0.7)

plt.xlim(0, 45)
plt.ylim(0, 0.10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', fontsize=10)
plt.show()
```

### Comparação dos Resultados Numéricos com o Gráfico e Interpretação

* **Área Azul Claro ($P(\chi^2 \le 10,0)$):** O cálculo numérico resultou em **3,18%**. No gráfico, essa região representa uma fatia muito fina na cauda inferior esquerda, coerente com a baixa probabilidade esperada de obtermos uma variação amostral tão diminuta por acaso.

* **Área Coral ($P(12,0 \le \chi^2 \le 28,0)$):** Corresponde à maior parte central da distribuição com **79,35%** de probabilidade total. Graficamente, a área coral cobre o pico central e a maior porção da curva de densidade, o que confirma a consistência matemática do cálculo.

* **Área Verde Claro ($P(\chi^2 > 32,0)$):** Reduzida ao extremo da cauda direita com valor calculado de **4,28%**. Como o gráfico mostra uma densidade muito baixa para valores de $\chi^2$ acima de 32, a representação visual da cauda verde está em perfeita harmonia com o resultado matemático estimado.

### Percentil na Distribuição Qui-quadrado

O **90º percentil** calculado para a distribuição Qui-quadrado com 20 graus de liberdade é de **28,412**.

# Distribuição F de Fisher-Snedecor
A Distribuição $F$ de Fisher-Snedecor é uma distribuição contínua de probabilidade que surge como a razão entre duas variáveis aleatórias independentes, cada uma seguindo uma distribuição Qui-Quadrado ($\chi^2$) dividida pelos respetivos graus de liberdade:$$F = \frac{U / d_1}{V / d_2}$$onde $U \sim \chi^2(d_1)$ e $V \sim \chi^2(d_2)$.Ao extrair amostras aleatórias e independentes de duas populações com distribuição normal e variâncias $\sigma_1^2$ e $\sigma_2^2$, a estatística de teste para a comparação de variâncias é definida por:$$F = \frac{S_1^2 / \sigma_1^2}{S_2^2 / \sigma_2^2}$$Sob a hipótese nula de igualdade de variâncias populacionais ($H_0: \sigma_1^2 = \sigma_2^2$), a expressão simplifica para a razão direta entre as duas variâncias amostrais:$$F = \frac{S_1^2}{S_2^2}$$1. Propriedades Matemáticas e TeoriaFunção Densidade de Probabilidade
/A densidade teórica da distribuição $F$ para $x > 0$ é dada por:$$f(x; d_1, d_2) = \frac{\left(\frac{d_1}{d_2}\right)^{\frac{d_1}{2}} x^{\frac{d_1}{2}-1}}{\text{B}\left(\frac{d_1}{2}, \frac{d_2}{2}\right) \left(1 + \frac{d_1}{d_2}x\right)^{\frac{d_1+d_2}{2}}}$$onde $\text{B}(a, b) = \frac{\Gamma(a)\Gamma(b)}{\Gamma(a+b)}$ representa a Função Beta e $\Gamma(\cdot)$ a Função Gama.O Princípio "Sinal / Ruído" e a Relação com a ANOVANa Análise de Variância (ANOVA), a estatística $F$ compara duas fontes independentes de variação:$$F = \frac{\text{Variância Entre Grupos (Sinal)}}{\text{Variância Dentro dos Grupos (Ruído)}}$$$F \approx 1$: A variação entre os grupos é da mesma magnitude do ruído aleatório amostral, não havendo evidência de efeito significativo do fator de agrupamento.$F \gg 1$: A variação explicada pelo fator de separação supera substancialmente a variação aleatória interna.Relação com o Teste $t$: Na comparação entre exatamente dois grupos independentes numa ANOVA de via única, cumpre-se rigidamente a relação $F = t^2$.Influência dos Graus de Liberdade na FormaGraus de Liberdade do Numerador ($d_1$): Se $d_1 \le 2$, a função de densidade é estritamente decrescente a partir da origem. Se $d_1 > 2$, a curva apresenta um valor modal único à esquerda de 1.Graus de Liberdade do Denominador ($d_2$): Controla a dispersão da cauda direita. Valores reduzidos de $d_2$ geram caudas pesadas. À medida que $d_2 \to \infty$, a variabilidade da estimativa do denominador diminui e a média da distribuição converge para $\frac{d_2}{d_2 - 2} \approx 1$.2. Aplicação Prática na Base de Dados de VinhosA divisão do conjunto de dados de vinhos brancos com base na variável Teor Alcoólico gera dois grupos de análise:Grupo de Baixo Teor ($\le 10\%$ vol.): $n_1 = 2007$ observações, com variância da acidez fixa $S_1^2 = 0,5981 \text{ (g/L)}^2$.Grupo de Alto Teor ($> 10\%$ vol.): $n_2 = 2780$ observações, com variância da acidez fixa $S_2^2 = 0,7684 \text{ (g/L)}^2$.Ao posicionar a maior variância no numerador, calcula-se a estatística $F$:$$F = \frac{S_2^2}{S_1^2} = \frac{0,7684}{0,5981} = 1,2847$$com graus de liberdade do numerador $d_1 = n_2 - 1 = 2779$ e do denominador $d_2 = n_1 - 1 = 2006$.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configuração estética
sns.set_theme(style="whitegrid")
plt.rcParams['font.size'] = 10

# 1. Carregamento e Filtragem dos Dados
df = pd.read_excel('/content/winequality-white.csv.xlsx')
df_clean = df[(df['alcohol'] <= 20) & (df['density'] <= 2) & (df['volatile acidity'] <= 2)].copy()

df_clean['grupo_alcool'] = np.where(df_clean['alcohol'] <= 10, 'Baixo Teor (<=10%)', 'Alto Teor (>10%)')

g_baixo = df_clean[df_clean['grupo_alcool'] == 'Baixo Teor (<=10%)']['fixed acidity']
g_alto = df_clean[df_clean['grupo_alcool'] == 'Alto Teor (>10%)']['fixed acidity']

v_baixo = g_baixo.var()
v_alto = g_alto.var()
n_baixo = len(g_baixo)
n_alto = len(g_alto)

# 2. Estatística F e Graus de Liberdade
f_obs = v_alto / v_baixo
d1 = n_alto - 1  # 2779
d2 = n_baixo - 1 # 2006
p_valor = stats.f.sf(f_obs, d1, d2)

# Probabilidades e Valores Críticos Teóricos
p_acumulada = stats.f.cdf(0.95, dfn=d1, dfd=d2)
p_intervalo = stats.f.cdf(1.05, dfn=d1, dfd=d2) - stats.f.cdf(0.95, dfn=d1, dfd=d2)
p_cauda_sup = stats.f.sf(1.0708, dfn=d1, dfd=d2)
percentil_95 = stats.f.ppf(0.95, dfn=d1, dfd=d2)

print(f"Grupo Baixo Teor: n = {n_baixo}, S1^2 = {v_baixo:.4f}")
print(f"Grupo Alto Teor:  n = {n_alto}, S2^2 = {v_alto:.4f}")
print(f"F Observado = {f_obs:.4f} | d1 = {d1}, d2 = {d2} | p-valor = {p_valor:.4e}")
print(f"P(F <= 0.95) = {p_acumulada:.4f}")
print(f"P(0.95 <= F <= 1.05) = {p_intervalo:.4f}")
print(f"P(F > 1.0708) = {p_cauda_sup:.4f}")
print(f"Percentil 95 (Valor Crítico alpha=0.05) = {percentil_95:.4f}")

# 3. Gráfico 1: Comparação Empírica dos Grupos
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x='grupo_alcool', y='fixed acidity', data=df_clean, ax=axes[0], palette=['#3498db', '#e74c3c'])
axes[0].set_title('Acidez Fixa por Grupo de Teor Alcoólico', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Grupo de Teor Alcoólico', fontsize=11)
axes[0].set_ylabel('Acidez Fixa (g/L)', fontsize=11)

sns.kdeplot(data=df_clean, x='fixed acidity', hue='grupo_alcool', common_norm=False, fill=True, alpha=0.3, palette=['#3498db', '#e74c3c'], ax=axes[1])
axes[1].set_title('Densidade Empírica da Acidez Fixa', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Acidez Fixa (g/L)', fontsize=11)

plt.tight_layout()
plt.show()

# 4. Gráfico 2: Densidade Teórica F com Áreas Sombreadas
fig, ax = plt.subplots(figsize=(10, 6))

x_f = np.linspace(0.8, 1.4, 1000)
pdf_f = stats.f.pdf(x_f, dfn=d1, dfd=d2)

ax.plot(x_f, pdf_f, color='#2c3e50', lw=2.5, label=f'Curva F(d1={d1}, d2={d2})')

# Áreas de Probabilidade
x_s1 = np.linspace(0.8, 0.95, 300)
ax.fill_between(x_s1, 0, stats.f.pdf(x_s1, dfn=d1, dfd=d2), color='#3498db', alpha=0.5, label=f'P(F <= 0.95) = {p_acumulada:.4f}')

x_s2 = np.linspace(0.95, 1.05, 300)
ax.fill_between(x_s2, 0, stats.f.pdf(x_s2, dfn=d1, dfd=d2), color='#2ecc71', alpha=0.5, label=f'P(0.95 <= F <= 1.05) = {p_intervalo:.4f}')

x_s3 = np.linspace(1.0708, 1.4, 300)
ax.fill_between(x_s3, 0, stats.f.pdf(x_s3, dfn=d1, dfd=d2), color='#e74c3c', alpha=0.5, label=f'P(F > 1.0708) = {p_cauda_sup:.4f}')

ax.axvline(f_obs, color='#8e44ad', linestyle='--', lw=2.5, label=f'F Observado = {f_obs:.4f}')
ax.axvline(percentil_95, color='#e74c3c', linestyle=':', lw=2, label=f'Valor Crítico (95%) = {percentil_95:.4f}')

ax.set_title('Distribuição F de Fisher e Regiões de Probabilidade', fontsize=12, fontweight='bold')
ax.set_xlabel('Razão de Variâncias F', fontsize=11)
ax.set_ylabel('Densidade de Probabilidade f(F)', fontsize=11)
ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

# Interpretação Probabilística
- Probabilidade Acumulada $P(F \le 0,95) = 0,1072$ ($10,72\%$): Representa a probabilidade de obter uma razão de variâncias amostrais menor ou igual a 0,95 quando as duas populações possuem variâncias rigorosamente iguais.
- Probabilidade no Intervalo $P(0,95 \le F \le 1,05) = 0,7726$ ($77,26\%$): Indica que $77,26\%$ dos pares de amostras extraídos de populações homocedásticas resultam numa razão $F$ contida no intervalo $[0,95; 1,05]$.
-  Probabilidade na Cauda Superior $P(F > 1,0708) = 0,0499 \approx 0,05$ ($5\%$): Corresponde ao nível de significância $\alpha = 0,05$, onde a probabilidade de obter uma razão $F$ superior a 1,0708 por puro acaso é de apenas $5\%$.
- Valor Crítico ($F_{0,95} = 1,0708$): Devido ao elevado número de graus de liberdade ($d_1 = 2779, d_2 = 2006$), a distribuição $F$ concentra-se fortemente em torno de 1. Qualquer valor amostral $F > 1,0708$ pertence à região de rejeição da hipótese nula.
- Conclusão Estatística: O valor observado de $F = 1,2847$ ultrapassa largamente o valor crítico de 1,0708, gerando um $p$-valor de $1,02 \times 10^{-9}$. Rejeita-se $H_0$, confirmando que os vinhos com teor alcoólico superior a $10\%$ possuem uma variabilida
