```python
import pandas as pd

file_path = '/content/dados_semana02.xlsx'

try:
    df = pd.read_excel(file_path)
    print(f"Successfully loaded '{file_path}' into a DataFrame.")
    print("Displaying all rows:")
    display(df)
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please make sure the filename is correct and the file exists.")
except Exception as e:
    print(f"An error occurred while reading the Excel file: {e}")
```

```python
import matplotlib.pyplot as plt

# Count the occurrences of each 'Doce favorito'
doce_counts = df['Doce favorito'].value_counts()

# Define a list of custom colors
custom_colors = ['#FF33CC', '#FF3300', '#9933FF', '#FFFF66'] # Example colors (light red, light blue, light green, light orange)

# Create the pie chart
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(doce_counts, labels=doce_counts.index, autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'black'}, colors=custom_colors)
ax.set_title('Preferência de doces dos clientes')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'Doce favorito' and sum 'Unidades compradas no ano'
units_by_sweet = df.groupby('Doce favorito')['Unidades compradas no ano'].sum().reset_index()

# Define a list of custom colors for the bar chart
custom_bar_colors = ['#FFFFCC', '#996633', '#FFCC99', '#CC9900'] # Example colors (blue violet, royal blue, crimson, light sea green)

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='Doce favorito', y='Unidades compradas no ano', data=units_by_sweet, palette=custom_bar_colors)
plt.title('Doces mais vendidos')
plt.xlabel('Doce Favorito')
plt.ylabel('Total de Unidades Compradas no Ano')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Idade'], palette='magma')
plt.title('Distribuição de Idades dos Clientes')
plt.ylabel('Idade')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the occurrences of each 'Doce favorito'
doce_counts_bar = df['Doce favorito'].value_counts().reset_index()
doce_counts_bar.columns = ['Doce favorito', 'Contagem']

# Create the bar chart for counts
plt.figure(figsize=(10, 6))
sns.barplot(x='Doce favorito', y='Contagem', data=doce_counts_bar, palette='viridis')
plt.title('Distribuição de Frequência dos Doces Favoritos')
plt.xlabel('Doce Favorito')
plt.ylabel('Número de Clientes')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['Unidades compradas no ano'], bins=10, kde=True, palette='viridis')
plt.title('Distribuição de Unidades Compradas no Ano')
plt.xlabel('Unidades Compradas no Ano')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
