##Pesquisa Nacional da Cesta Básica de Alimentos

Realizada pelo DIEESE (Departamento Intersindical de Estatística e Estudos Socioeconômicos), é um levantamento mensal realizado nas 27 capitais do Brasil que acompanha o preço de um conjunto de produtos alimentícios essenciais e permite a comparação de custos dos principais alimentos básicos consumidos pelos brasileiros.

Os itens básicos pesquisados foram definidos pelo Decreto Lei nº 399, de 30 de abril de 1938, que regulamentou o salário mínimo no Brasil e está vigente até os dias atuais. O Decreto determinou que a cesta de alimentos fosse composta por 13 produtos alimentícios em quantidades suficientes para garantir, durante um mês, o sustento e bem-estar de um trabalhador em idade adulta. Os bens e quantidades estipuladas foram diferenciados por região, variando conforme os hábitos de consumo.

Itens avaliados: A cesta acompanha 13 produtos básicos definidos por um decreto de 1938 (como carne, leite, feijão, arroz, farinha, pão, café, óleo, manteiga, açúcar, batata, tomate e fruta).

Para esse trabalho foi considerado os dados da cidade de São Paulo no período entre Janeiro/2020 à Julho/2026.

#Tabela de provisões mínimas estipuladas pelo Decreto Lei n° 399

Carne: 6,0 kg

Leite: 7,5 litros

##Métricas utilizadas:

Valor Final da Cesta:

O preço médio ponderado de cada um dos 13 produtos é multiplicado pela quantidade prevista em decreto. A somatória desses valores resulta no custo total mensal da cesta básica daquela cidade.

Tempo de Trabalho Necessário:

Este indicador mede quantas horas o cidadão que recebe um salário mínimo precisa trabalhar no mês exclusivamente para pagar pela sua alimentação básica.O cálculo adota a jornada de trabalho padrão estipulada pela Constituição Federal, que é de 220 horas mensais e divide-se o valor total da cesta básica pelo valor do salário mínimo nominal vigente. Em seguida, multiplica-se o resultado por 220.

Impacto Financeiro (Comprometimento do Salário):

O impacto financeiro aponta a porcentagem da renda líquida do trabalhador absorvida pela compra dos alimentos da cesta.
Antes de calcular o impacto, o DIEESE deduz do salário mínimo bruto o percentual de 7,5%, referente à contribuição obrigatória da Previdência Social (INSS), obtendo o salário mínimo líquido.
O custo da cesta básica é dividido pelo salário mínimo líquido, e o resultado é multiplicado por 100 para gerar o percentual.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define the file path as specified by the user
file_path = '/content/dados-seminario .xls'

# Check if the file exists
if not os.path.exists(file_path):
    print(f"Error: The file '{file_path}' was not found.")
    print("Please ensure the Excel file 'dados-seminario .xls' is uploaded to the Colab environment in the '/content/' folder.")
    print("You can upload it using the folder icon on the left in Colab and dragging the file to the '/content/' folder.")
    raise FileNotFoundError(f"Required file not found: {file_path}")

# Load the Excel file and extract 'Preço Médio' sheet
excel_file = pd.ExcelFile(file_path)
all_sheets_df = {}
for sheet_name in excel_file.sheet_names:
    all_sheets_df[sheet_name] = excel_file.parse(sheet_name)

df_preco_medio = all_sheets_df['Preço Médio'].copy()
df_preco_medio['Data'] = pd.to_datetime(df_preco_medio['Data'], format='%m-%Y')
df_preco_medio = df_preco_medio.set_index('Data')

# Extract year and month
df_preco_medio['Ano'] = df_preco_medio.index.year
df_preco_medio['Mês'] = df_preco_medio.index.month

# Calculate yearly average for 'Total da Cesta'
yearly_avg = df_preco_medio.groupby('Ano')['Total da Cesta'].mean().reset_index()
yearly_avg = yearly_avg.rename(columns={'Total da Cesta': 'Média Anual da Cesta'})

# Merge yearly average back to the main DataFrame
df_plot = pd.merge(df_preco_medio.reset_index(), yearly_avg, on='Ano', how='left')

# Calculate deviation from yearly average
df_plot['Desvio da Média Anual'] = df_plot['Total da Cesta'] - df_plot['Média Anual da Cesta']

# Filter for years 2020-2026
df_plot_filtered = df_plot[(df_plot['Ano'] >= 2020) & (df_plot['Ano'] <= 2026)]

# Plotting the deviation as a bar chart
plt.figure(figsize=(16, 8))

# Define colors for each year
unique_years = sorted(df_plot_filtered['Ano'].unique())
colors = sns.color_palette('viridis', n_colors=len(unique_years))
year_to_color = dict(zip(unique_years, colors))

# Plot bars for each month, colored by year
plt.bar(df_plot_filtered['Data'], df_plot_filtered['Desvio da Média Anual'],
        width=pd.Timedelta(days=20), # Adjust width for better visualization of monthly bars
        color=df_plot_filtered['Ano'].map(year_to_color))

plt.title('Desvio do Preço Total da Cesta da Média Anual (2020-2026)', fontsize=16)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Desvio (R$)', fontsize=12)
plt.axhline(0, color='red', linestyle='--', linewidth=0.8) # Add a zero line for reference
plt.grid(True, linestyle='--', alpha=0.7)

# Create a legend manually for the colors
handles = [plt.Rectangle((0,0),1,1, color=year_to_color[year]) for year in unique_years]
plt.legend(handles, unique_years, title='Ano', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read all sheets from the Excel file
excel_file = pd.ExcelFile('/content/dados-seminario .xls') # Corrected filename
all_sheets_df = {}
for sheet_name in excel_file.sheet_names:
    all_sheets_df[sheet_name] = excel_file.parse(sheet_name)

df_tempo_trabalho = all_sheets_df['Tempo de trabalho'].copy()

# Convert 'Data' column to datetime objects
df_tempo_trabalho['Data'] = pd.to_datetime(df_tempo_trabalho['Data'], format='%m-%Y')

# Define a function to convert time strings (e.g., '109h35m') to total hours (float)
def convert_time_to_hours(time_str):
    if pd.isna(time_str):
        return None
    if 'h' not in time_str:
        return float(time_str) # Assume it's already hours if no 'h'
    parts = time_str.replace('m', '').split('h')
    hours = float(parts[0])
    minutes = float(parts[1]) if parts[1] else 0
    return hours + minutes / 60

# Apply the conversion function to all relevant columns
# Exclude 'Data' from conversion
columns_to_convert = [col for col in df_tempo_trabalho.columns if col != 'Data']
for col in columns_to_convert:
    df_tempo_trabalho[col] = df_tempo_trabalho[col].apply(convert_time_to_hours)

# Set 'Data' as index for easier plotting
df_tempo_trabalho = df_tempo_trabalho.set_index('Data')

# Select actual columns to plot, based on the dataframe content
columns_to_plot_tempo = ['Total da Cesta', 'Carne', 'Leite', 'Arroz', 'Manteiga']

# Filter out columns that might not exist (though they should now)
existing_columns_tempo = [col for col in columns_to_plot_tempo if col in df_tempo_trabalho.columns]

plt.figure(figsize=(15, 7))
sns.lineplot(data=df_tempo_trabalho[existing_columns_tempo], dashes=False)
plt.title('Comparativo: Tempo de Trabalho para Cesta Básica e Componentes ao Longo do Tempo', fontsize=16)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Tempo de Trabalho (Horas)', fontsize=12)
plt.legend(title='Componente', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Ensure dataframes are loaded from the Excel file if not already in memory
if 'all_sheets_df' not in locals():
    file_path = '/content/dados-seminario .xls'
    if not os.path.exists(file_path):
        print(f"Error: The file '{file_path}' was not found.")
        raise FileNotFoundError(f"Required file not found: {file_path}")
    excel_file = pd.ExcelFile(file_path)
    all_sheets_df = {}
    for sheet_name in excel_file.sheet_names:
        all_sheets_df[sheet_name] = excel_file.parse(sheet_name)

# --- Prepare df_preco_medio (Cesta Básica in R$) ---
df_preco_medio = all_sheets_df['Preço Médio'].copy()
df_preco_medio['Data'] = pd.to_datetime(df_preco_medio['Data'], format='%m-%Y')
df_preco_medio = df_preco_medio.set_index('Data')

# --- Prepare df_tempo_trabalho (Tempo de trabalho for Cesta Básica) ---
df_tempo_trabalho = all_sheets_df['Tempo de trabalho'].copy()
df_tempo_trabalho['Data'] = pd.to_datetime(df_tempo_trabalho['Data'], format='%m-%Y')

# Function to convert time strings (e.g., '109h35m') to total hours (float)
def convert_time_to_hours(time_str):
    if pd.isna(time_str):
        return None
    if isinstance(time_str, (int, float)):
        return float(time_str)
    if 'h' not in time_str:
        return float(time_str)
    parts = time_str.replace('m', '').split('h')
    hours = float(parts[0])
    minutes = float(parts[1]) if parts[1] else 0
    return hours + minutes / 60

# Apply the conversion to 'Total da Cesta' in df_tempo_trabalho
df_tempo_trabalho['Total da Cesta (Horas)'] = df_tempo_trabalho['Total da Cesta'].apply(convert_time_to_hours)
df_tempo_trabalho = df_tempo_trabalho.set_index('Data')

# --- Calculate Salário Mínimo Nominal ---
# Using the formula: Tempo de Trabalho = (Cesta Básica / Salário Mínimo Nominal) * 220
# So, Salário Mínimo Nominal = (Cesta Básica / Tempo de Trabalho) * 220

# Merge df_preco_medio and df_tempo_trabalho on their 'Data' index
merged_df = pd.merge(df_preco_medio[['Total da Cesta']],
                     df_tempo_trabalho[['Total da Cesta (Horas)']],
                     left_index=True, right_index=True, how='inner')

merged_df['Salário Mínimo Nominal'] = (merged_df['Total da Cesta'] / merged_df['Total da Cesta (Horas)']) * 220

# --- Filter for the period 2020-2026 ---
merged_df = merged_df[(merged_df.index.year >= 2020) & (merged_df.index.year <= 2026)]

# --- Index to 100 based on January 2020 ---
jan_2020_data = merged_df.loc['2020-01-01']

base_cesta = jan_2020_data['Total da Cesta']
base_salario = jan_2020_data['Salário Mínimo Nominal']

merged_df['Cesta Básica Indexada'] = (merged_df['Total da Cesta'] / base_cesta) * 100
merged_df['Salário Mínimo Nominal Indexado'] = (merged_df['Salário Mínimo Nominal'] / base_salario) * 100

# --- Plotting the indexed values with fill_between ---
plt.figure(figsize=(16, 8))

plt.plot(merged_df.index, merged_df['Cesta Básica Indexada'], label='Cesta Básica (Indexada)', color='blue', linewidth=2)
plt.plot(merged_df.index, merged_df['Salário Mínimo Nominal Indexado'], label='Salário Mínimo Nominal (Indexado)', color='green', linewidth=2)

# Fill between the lines to highlight the gap (defasagem)
# Use a condition to color based on whether the basket is above or below the minimum wage
plt.fill_between(merged_df.index, merged_df['Cesta Básica Indexada'], merged_df['Salário Mínimo Nominal Indexado'],
                 where=(merged_df['Cesta Básica Indexada'] > merged_df['Salário Mínimo Nominal Indexado']),
                 facecolor='red', alpha=0.3, label='Defasagem: Cesta > Salário')
plt.fill_between(merged_df.index, merged_df['Cesta Básica Indexada'], merged_df['Salário Mínimo Nominal Indexado'],
                 where=(merged_df['Cesta Básica Indexada'] < merged_df['Salário Mínimo Nominal Indexado']),
                 facecolor='green', alpha=0.3, label='Defasagem: Cesta < Salário')


plt.title('Cesta Básica vs. Salário Mínimo Nominal (Jan/2020 = 100) com Defasagem Acumulada', fontsize=16)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Valor Indexado (Jan/2020 = 100)', fontsize=12)
plt.axhline(100, color='grey', linestyle='--', linewidth=0.8, label='Base Jan/2020 = 100')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

Devido à sua alta representatividade (18,5%) e forte correlação ($0,90$), choque nos preços do café é um dos principais desencadeadores de surtos inflacionários na cesta.

##Resiliência e Previsibilidade dos Alimentos Básicos
Os itens tradicionais do prato brasileiro — Arroz (2,0%), Feijão (3,3%), Leite (2,7%) e Açúcar (1,8%) — apresentam custos médios reduzidos (abaixo de R$ 8,00), baixa volatilidade ao longo do tempo e representam, somados, menos de 10% do valor final da cesta. Suas dinâmicas dependem mais de safras específicas do que da tendência geral de mercado.
