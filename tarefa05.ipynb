```python
import pandas as pd

# Load the Excel file into a pandas DataFrame
df = pd.read_excel('usda_data.xlsx')

# Display the first 5 rows of the DataFrame
display(df)
```

```python
# Calculate mean and median for 'Proteína (g)'
protein_mean = df['Proteína (g)'].mean()
protein_median = df['Proteína (g)'].median()

print(f"Média de Proteína (g): {protein_mean:.2f}")
print(f"Mediana de Proteína (g): {protein_median:.2f}")

# Calculate mean and median for 'Sódio (mg)'
sodium_mean = df['Sódio (mg)'].mean()
sodium_median = df['Sódio (mg)'].median()

print(f"\nMédia de Sódio (mg): {sodium_mean:.2f}")
print(f"Mediana de Sódio (mg): {sodium_median:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plotting the distribution of 'Proteína (g)'
plt.figure(figsize=(10, 6))
sns.histplot(df['Proteína (g)'], kde=True, bins=15)
plt.title('Distribuição do Teor de Proteína (g)')
plt.xlabel('Proteína (g)')
plt.ylabel('Frequência')
plt.axvline(df['Proteína (g)'].mean(), color='r', linestyle='--', label=f'Média: {df["Proteína (g)"].mean():.2f}')
plt.axvline(df['Proteína (g)'].median(), color='g', linestyle='-', label=f'Mediana: {df["Proteína (g)"].median():.2f}')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gerando o boxplot horizontal para 'Proteína (g)'
plt.figure(figsize=(8, 6))
sns.boxplot(x=df['Proteína (g)'])
plt.title('Teor de Proteína (g)')
plt.xlabel('Proteína (g)')
plt.ylabel('') # Remove o rótulo do eixo Y para boxplot horizontal
plt.grid(axis='x', alpha=0.75)
plt.show()
```

```python
# Calcule a moda de 'Proteína (g)'
protein_mode = df['Proteína (g)'].mode()

print(f"Moda de Proteína (g):\n{protein_mode}")
```

```python
# Plotting the distribution of 'Sódio (mg)'
plt.figure(figsize=(10, 6))
sns.histplot(df['Sódio (mg)'], kde=True, bins=20)
plt.title('Distribuição do Teor de Sódio (mg)')
plt.xlabel('Sódio (mg)')
plt.ylabel('Frequência')
plt.axvline(df['Sódio (mg)'].mean(), color='r', linestyle='--', label=f'Média: {df["Sódio (mg)"].mean():.2f}')
plt.axvline(df['Sódio (mg)'].median(), color='g', linestyle='-', label=f'Mediana: {df["Sódio (mg)"].median():.2f}')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```
