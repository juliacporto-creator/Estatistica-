```python
import pandas as pd

# Read the Excel file into a pandas DataFrame
df_chocolate = pd.read_excel('/content/dados chocolate ideal.xlsx')

# Display the first 5 rows of the DataFrame
display(df_chocolate)
```

```python
# Calculate the simple frequency for 'Tipo de chocolate'
frequency_table = df_chocolate['Tipo de chocolate'].value_counts()

# Display the frequency table
display(frequency_table)
```

```python
# Calculate the total number of observations
total_observations = frequency_table.sum()

# Calculate the relative frequency
relative_frequency_table = (frequency_table / total_observations) * 100

# Display the relative frequency table
display(relative_frequency_table)
```

```python
# Calculate the absolute frequency for 'Quantas barras de\nChocolate come por semana'
frequency_barras_semana = df_chocolate['Quantas barras de\nChocolate come por semana'].value_counts()

# Display the frequency table
display(frequency_barras_semana)
```

```python
total_observacoes_barras = frequency_barras_semana.sum()
relative_frequency_barras = (frequency_barras_semana / total_observacoes_barras) * 100

display(relative_frequency_barras)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a bar chart for 'Tipo de chocolate'
plt.figure(figsize=(10, 6))
sns.barplot(x=frequency_table.index, y=frequency_table.values, palette='viridis')
plt.title('Preferência de Tipos de Chocolate')
plt.xlabel('Tipo de Chocolate')
plt.ylabel('Número de Pessoas')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout()
plt.show()
```
